Classical Bit: The basic logical unit for computer science, storing information as a binary number (either 1 or 0). The space of states is a simple mathematical set: {0.1}.

Qubit: The quantum version of a two-state system. The state vector is a vector space with infinitely possible states, requiring linear algebra to to describe their transformations.

Quantum States: A single bit is written as $\psi = \alpha\vert{}0\rangle + \beta\vert{}1\rangle$ . The coefficients $\alpha$ and $\beta$ represent the probability of the qubit collapsing into that respective state upon measurement. By this, they capture a large amount of information as the number of possible states increases. 

1-Qubit SystemIn bra-ket notation, the state is a linear combination of two basis states:
  $$\vert{}\psi\rangle = \alpha\vert{}0\rangle + \beta\vert{}1\rangle$$As a column vector, this translates to a $2 \times 1$ matrix:$$\vert{}\psi\rangle = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}$$2-Qubit SystemIn bra-ket notation, the state expands to a linear combination of four basis states:
  $$\vert{}\psi\rangle = \alpha_{00}\vert{}00\rangle + \alpha_{01}\vert{}01\rangle + \alpha_{10}\vert{}10\rangle + \alpha_{11}\vert{}11\rangle$$
  First, each individual basis state is written as a standard column vector:
  $$\vert{}00\rangle = \begin{pmatrix} 1 \\ 0 \\ 0 \\ 0 \end{pmatrix}$$

 $$\vert{}01\rangle = \begin{pmatrix} 0 \\ 1 \\ 0 \\ 0 \end{pmatrix}$$

 $$\vert{}10\rangle = \begin{pmatrix} 0 \\ 0 \\ 1 \\ 0 \end{pmatrix}$$

 $$\vert{}11\rangle = \begin{pmatrix} 0 \\ 0 \\ 0 \\ 1 \end{pmatrix}$$

 When you multiply each basis vector by its respective probability amplitude ($\alpha$) and add them together, the full mathematical equation expands to:$$\vert{}\psi\rangle = \alpha_{00} \begin{pmatrix} 1 \\ 0 \\ 0 \\ 0 \end{pmatrix} + \alpha_{01} \begin{pmatrix} 0 \\ 1 \\ 0 \\ 0 \end{pmatrix} + \alpha_{10} \begin{pmatrix} 0 \\ 0 \\ 1 \\ 0 \end{pmatrix} + \alpha_{11} \begin{pmatrix} 0 \\ 0 \\ 0 \\ 1 \end{pmatrix}$$
 
  
  Which simplifies perfectly into a $4 \times 1$ state vector matrix:$$\vert{}\psi\rangle = \begin{pmatrix} \alpha_{00} \\ \alpha_{01} \\ \alpha_{10} \\ \alpha_{11} \end{pmatrix}$$


The Quantum state: 
$\psi$(Psi): This is the "ket" representing the total, overall state of your 2-qubit system.

Superposition: The + signs represent the system not being in 1 state. It is a linear combination of all 4 possible states.

The Basis Vectors:
$\vert{}00\rangle$, $\vert{}01\rangle$, $\vert{}10\rangle$, $\vert{}11\rangle$. These are the computational basis states.Think of them as x,y,z,w axis of a 4 dimensional vector space. They represent the four measurable outcomes if you were to look at the system.

The Probability Amplitudes:
$\alpha_{00}$, $\alpha_{01}$, $\alpha_{10}$, $\alpha_{11}$: These are scalar coefficients (specifically, complex numbers) attached to each basis state.

Physical Meaning:n A-level physics, you learned that the square of a wave function's amplitude gives the probability of finding a particle in a certain place. The exact same rule applies here. If you measure this 2-qubit system, the probability of it collapsing into the $\vert{}00\rangle$ state is exactly $\vert{}\alpha_{00}\vert{}^2$.Because the system must collapse into one of these four states upon measurement, the sum of all their probabilities must equal 100%. Mathematically, this means:$$\vert{}\alpha_{00}\vert{}^2 + \vert{}\alpha_{01}\vert{}^2 + \vert{}\alpha_{10}\vert{}^2 + \vert{}\alpha_{11}\vert{}^2 = 1$$



Matrix Multiplication: Transformations are achieved by multiplying the state vector by a matrix. For example, a bit flip gate (Pauli-X) translates to $\vert{}0\rangle$ to $\vert{}1\rangle$

$$\begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix} \begin{pmatrix} 1 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \\ 1 \end{pmatrix}$$
